```python
from fastai.tabular.all import *
import pandas as pd
path = Path('.')
df = pd.read_csv(path/'Loan_Default.csv')
cat_names = ['loan_limit', 'Gender', 'approv_in_adv', 'loan_type', 'loan_purpose',
                                             'Credit_Worthiness', 'open_credit', 'business_or_commercial','Neg_ammortization',
                                             'interest_only', 'lump_sum_payment', 'construction_type', 'occupancy_type','Secured_by', 'total_units',
                                             'credit_type', 'co-applicant_credit_type', 'age', 'submission_of_application', 'Region', 'Security_Type',
                                             ]
cont_names = [ x for x in df.columns if x not in cat_names and x!='Status']
cont_names.remove('ID')
cont_names
```

```python
dls = TabularDataLoaders.from_csv(path/'Loan_Default.csv', path=path, bs=64, y_names='Status',
                                  cat_names=cat_names, cont_names=cont_names,
                                  procs=[Categorify, FillMissing, Normalize]
                                  )

learn = tabular_learner(dls, metrics=accuracy)
learn.fit_one_cycle(2)
```

```python
learn.show_results(max_n=64)
```
